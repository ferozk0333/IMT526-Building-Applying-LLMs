```python
import pandas as pd
```

```python
df = pd.read_csv('/content/100_Unique_QA_Dataset.csv')
df.head()
```

```python
# Tokenize
def tokenize(text):
  text = text.lower()
  text = text.replace('?','')
  text = text.replace("'",'')
  return text.split()
```

```python
# Vocab
vocab = {
    '<OOV>':0
}
```

```python
def build_vocab(row):

  tokenized_ques = tokenize(row['question'])
  tokenized_ans = tokenize(row['answer'])
  merged_tokens = tokenized_ques + tokenized_ans

  for token in merged_tokens:
    if token not in vocab:
      vocab[token] = len(vocab)
```

```python
df.apply(build_vocab, axis = 1)
```

```python
# Convert text to numerical indices
def text_to_index(text, vocab):

  text_indiced = []
  for token in tokenize(text):
    if token in vocab:
      text_indiced.append(vocab[token])
    else:
      text_indiced.append(vocab['<OOV>'])

  return text_indiced
```

```python
text_to_index('What is campusX', vocab)
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
```

```python
class QADataset(Dataset):
  def __init__(self, df, vocab):
    self.df = df
    self.vocab = vocab

  def __len__(self):
    return self.df.shape[0]

  def __getitem__(self, index):
    numerical_ques = text_to_index(df.iloc[index]['question'], self.vocab)
    numerical_ans = text_to_index(df.iloc[index]['answer'], self.vocab)

    return torch.tensor(numerical_ques), torch.tensor(numerical_ans)
```

```python
dataset = QADataset(df, vocab)
```

```python
dataset[5]
```

```python
dataloader = DataLoader(dataset, batch_size=1, shuffle=True, )
```

```python
for question, answer in dataloader:
  print(question, answer)
```

```python
## Neural Network
import torch.nn as nn
import torch.optim as optim
```

```python
class SimpleRNN(nn.Module):

  def __init__(self, vocab_size):
    super().__init__()

    self.embedding = nn.Embedding(num_embeddings=vocab_size, embedding_dim=50)
    self.rnn = nn.RNN(50, 64, batch_first=True)
    self.fc = nn.Linear(64,vocab_size)


  def forward(self, question):
    embedded_question = self.embedding(question)
    hidden, final = self.rnn(embedded_question)
    output = self.fc(final.squeeze(0))

    return output
```

```python
learning_rate = 0.001
epochs = 100
```

```python
model = SimpleRNN(len(vocab))
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = learning_rate)
```

```python
# Training Loop

for epoch in range(epochs):
  total_epoch_loss = 0.0
  for question, answer in dataloader:

    optimizer.zero_grad()

    #Forward pass
    output = model(question)

    #Loss
    loss = criterion(output, answer[0])

    #Backward pass
    loss.backward()

    #Update params
    optimizer.step()

    total_epoch_loss += loss

  print(f"Epoch: [{epoch+1}/{epochs}] ---> Loss: {total_epoch_loss:.4f}")
```

```python
index_to_word = list(vocab.keys())
```

```python
# Prediction
def predict(model, question, threshold = 0.5):

  # Convert Question to numbers [1,4,2,6]
  numerical_question = text_to_index(question, vocab)

  # Convert to tensors
  question_tensor = torch.tensor(numerical_question).unsqueeze(0)  # [3] --> [1,3]

  # Send to model
  output = model(question_tensor)

  # Convert logits to probs
  probs = torch.nn.functional.softmax(output, 1)

  # Find index of max prob
  value, index = torch.max(probs, dim = 1)


  if value < threshold:
    return "I don't know"

  else:
    return index_to_word[index]
```

```python
predict(model, "What is the Capital of France?")
```
